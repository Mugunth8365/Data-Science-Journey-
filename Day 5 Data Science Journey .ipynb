```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

players = pd.DataFrame({
    "player_id": [1, 2, 3, 4, 5],
    "name": ["Arun", "Bala", "Chitra", "Dev", "Esha"],
    "team_id": [10, 10, 20, 30, 40]
})

teams = pd.DataFrame({
    "team_id": [10, 20, 30],
    "team_name": ["Lions", "Tigers", "Bears"],
    "city": ["Chennai", "Mumbai", "Delhi"]
})

scores = pd.DataFrame({
    "player_id": [1, 1, 2, 3, 3, 3, 6],
    "runs": [45, 60, 30, 80, 25, 55, 40]
})
```

```python
print(players.merge(scores,on="player_id"))

left1=pd.DataFrame(players.merge(scores,on="player_id",how="left"))
print(left1)
print(left1[left1["runs"].isna()])

left2=pd.DataFrame(players.merge(teams,on="team_id",how="left"))
print(left2)
print(left2[left2["team_name"].isnull()])

left3=pd.DataFrame(left1.merge(teams,on="team_id",how="left"))
print(left3)
print(left3.groupby("team_name")["runs"].sum())

print(left3.groupby("city")["runs"].mean())

print(left3[left3["runs"].isnull()])

sns.barplot(x="team_name",y="runs",data=left3,estimator=sum)
plt.title("Total runs scored by each team")
plt.xlabel("Team Name")
plt.ylabel("Total Runs")
plt.show()





```

```python
total_score= left3.groupby("name").sum()["runs"]
print(total_score)
print(f"The top run scorer is {total_score.idxmax()} with {int(total_score.max())} runs.")
```

An inner merge keeps only the rows that have a match in both tables, while a left merge keeps every row of the left table and fills the unmatched columns with NaN.
