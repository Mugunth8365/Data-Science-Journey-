```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
```

```python
print(df)
```

```python
print(df.info())
```

```python
print(df.isna().sum())
```

```python
print(df.duplicated().sum())
```

```python
df["embark_town"]=df["embark_town"].fillna(df["embark_town"].value_counts().idxmax())
print(df.isna().sum())
print(df["embark_town"])
```

```python
a=df.shape
print(a)
df=df.drop_duplicates()
print(df.shape)


```

```python
print(df["age"].isna().sum())
print(df.info())
df["is_child"]=df["age"]<16
print(df["is_child"].shape)
print(df["is_child"].sum())
print(df["is_child"].info())
```

```python
print(df.info())
df=df.drop(columns=["deck"])
print(df.info())
```

```python
sns.barplot(x="sex",y="survived",data=df)
plt.xlabel("Sex")
plt.ylabel("Survived")
plt.title("Sex vs Survived")
plt.show()
a,b=df.groupby("sex")["survived"].mean()*100
print(a,b)
print(f"Female {a}% has bigger survival rate compared to Male{b}%")
```
