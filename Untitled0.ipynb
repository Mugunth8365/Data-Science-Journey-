```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")   # restaurant bills dataset
df.head()                       # see the first 5 rows
```

```python
df.shape            # rows, columns
df.info()           # column types, missing values
df.describe()       # mean, min, max, etc.
```

```python
df["tip"].mean()                       # average tip
df.groupby("day")["total_bill"].mean() # avg bill per day
```

```python
print(df.groupby(df["smoker"] == "Yes")["tip"].mean())
```
